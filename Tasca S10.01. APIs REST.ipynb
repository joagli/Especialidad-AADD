1. Consulta l’API pública JSONPlaceholder utilitzant el mètode GET per obtenir:

Llista de publicacions (/posts)
Llista d’usuaris (/users)
Llista de tasques (/todos)


```python
import requests
import pandas as pd

response = requests.get("https://jsonplaceholder.typicode.com/posts")
print(response.status_code)
print()
print(response.json())

response = requests.get("https://jsonplaceholder.typicode.com/users")
print(response.status_code)
print()
print(response.json())

response = requests.get("https://jsonplaceholder.typicode.com/todos")
print(response.status_code)
print()
print(response.json())
```

2. Mostra per pantalla:

La quantitat total de cada recurs.
El codi d'estat de cada petició.

```python
response = requests.get("https://jsonplaceholder.typicode.com/posts")
print("Status code: ", response.status_code,)
print()
data = response.json()
print("Cantidad total: ", len(data))
print()

response = requests.get("https://jsonplaceholder.typicode.com/users")
print("Status code: ", response.status_code)
print()
data = response.json()
print("Cantidad total: ", len(data))
print()

response = requests.get("https://jsonplaceholder.typicode.com/todos")
print("Status code: ", response.status_code)
print()
data = response.json()
print("Cantidad total: ", len(data))
```

3. Fes una petició a una publicació inexistent per obtenir un error 404 i mostra el codi d'estat rebut.

```python
response = requests.get("https://jsonplaceholder.typicode.com/posts/9999")

print(response.status_code)
```

4. Fes una petició POST per crear una nova publicació fictícia. Inclou el títol, el cos del missatge i un userId.

Mostra la resposta JSON.
Mostra el codi d'estat.

```python
new_post = {'userId': 101, 'title': 'sunt aut', 'body': 'quia et suscipit'}
r = requests.post("https://jsonplaceholder.typicode.com/posts", json=new_post)
print(r.status_code)
print()
print(r.json())
```

5. Fes una petició PATCH per modificar parcialment una publicació existent.

Mostra la resposta JSON.
Mostra el codi d'estat.

```python
r = requests.patch("https://jsonplaceholder.typicode.com/posts/100", json={"title": "El nuevo título"})
print(r.status_code)
print()
print(r.json())
```

6. Fes una petició DELETE sobre una publicació.

Mostra la resposta JSON.
Mostra el codi d'estat.

```python
r = requests.delete("https://jsonplaceholder.typicode.com/posts/95")
print(r.status_code)
print()
print(r.json())
```
