```python
candidats = []

def obtenir_dades(test_data=None): #Si recibe una lista se genera el iterador, si no, comienza con el registro manual
  source = iter(test_data) if test_data else None

  while len(candidats) < 10:
    if source:
      try:
        candidat = {
            "nombre": next(source),
            "telefono": next(source),
            "exp_laboral": next(source),
            "catalan": next(source),
            "excel": next(source),
            "titulo": next(source)}
      except StopIteration:
          print("No hay más datos en la lista") #En caso que la lista de datos tenga menos de 10 candidatos con todos sus datos
          break
    else:
      candidat = {
          "nombre": input("Ingresa tu nombre y apellidos: "),
          "telefono": input("Teléfono: "),
          "exp_laboral": input("Años de experiencia laboral: "),
          "catalan": input("Conocimiento de catalán (0 = sin conocimiento, 1 = hasta nivel B1, 2 = nivel B2, 3 = C1 o superior): "),
          "excel": input("Dominio de Excel (0 = nivel básico, 1 = uso de funciones, 2 = tablas dinámicas, 3 = Visual Basic y Macros): "),
          "titulo": input("Título académico (0 = ESO, 1 = FP, 2 = Universidad, 3 = Master o superior): ")}

    if not validar_dades(candidat):
      print(f"Datos inválidos de {candidat["nombre"]}. Registro no realizado.")
      continue

    candidat["puntuacio"] = calcular_puntuacio(candidat)
    candidat["clasificacio"] = clasificar_candidat(candidat["puntuacio"])
    candidats.append(candidat)
    mostrar_individual(candidat, candidat["puntuacio"], candidat["clasificacio"])
```

```python
def validar_dades(candidat):
  try:
    exp = int(candidat["exp_laboral"])
    if exp < 0:
      return False

    catalan = int(candidat["catalan"])
    if catalan not in [0, 1, 2, 3]:
      return False

    excel = int(candidat["excel"])
    if excel not in [0, 1, 2, 3]:
      return False

    titulo = int(candidat["titulo"])
    if titulo not in [0, 1, 2, 3]:
      return False

  except ValueError:
      return False

  return True

print(validar_dades({"nombre": "Test", "telefono": "600", "exp_laboral": 5, "catalan": 3, "excel": 2, "titulo": 1}))
print(validar_dades({"nombre": "Error", "telefono": "600", "exp_laboral": 5, "catalan": 7, "excel": 2, "titulo": 1}))
```

```python
def calcular_puntuacio(candidat):
  exp = float(candidat["exp_laboral"] * 1.5)
  catalan = int(candidat["catalan"])
  excel = int(candidat["excel"])
  titulo = int(candidat["titulo"])
  return exp + catalan + excel + titulo

print(calcular_puntuacio({"exp_laboral": 6, "catalan": 3, "excel": 3, "titulo": 3}))
print(calcular_puntuacio({"exp_laboral": 1, "catalan": 2, "excel": 1, "titulo": 0}))
```

```python
def clasificar_candidat(puntuacio):
  if puntuacio >= 12:
    return "Prioritat Alta"
  elif 8 <= puntuacio < 12:
    return "Candidat adequat"
  else:
    return "No prioritari"

print(clasificar_candidat(18.0))
print(clasificar_candidat(9.5))
print(clasificar_candidat(4.5))
```

```python
def procesar_clients(clients):
  clients_procesats = []

  for client in clients:
    info_segment = obtenir_segment_i_factor(client["codi_comportament"])
    factor_edat = obtenir_factor_edat(client["grup_edat"])
    risc_final = calcular_risc(info_segment["factor_codi"], factor_edat)

    client_procesat = client.copy() # Copia del diccionario original para no modificarlo.
    client_procesat["segment"] = info_segment["segment"]
    client_procesat["missatge"] = info_segment["missatge"]
    client_procesat["factor_codi"] = info_segment["factor_codi"]
    client_procesat["factor_edat"] = factor_edat
    client_procesat["risc_final"] = risc_final

    clients_procesats.append(client_procesat)

  return clients_procesats
```

```python
def parsear_paciente(text):
  nombre, edad, sintomas_origen = text.split("|")
  sintomas = [s.strip().lower().replace(" ", "_")
              for s in sintomas_origen.split(",")]

  return {"nombre": nombre,
          "edad": edad,
          "sintomas": sintomas}

print(parsear_paciente("Maria|45|tos, Fiebre Alta, dolor_pecho"))
```

```python
pacientes = [
    "Maria|45|tos, Fiebre Alta, dolor_pecho",
    "Luis|33|dolor_cabeza,faTiga",
    "Sara|67|fiebre alta, dificultad RESPIRAR, tos",
    "Jordi|52|fatiga , tos",
    "Anna|29|fiebre alta,dolor_cabeza"
]

niveles_riesgo = {
    "fiebre_alta": 3,
    "dificultad_respirar": 5,
    "dolor_pecho": 4,
    "tos": 1,
    "fatiga": 1,
    "dolor_cabeza": 1
}

p = parsear_paciente("Maria|45|tos, Fiebre Alta, dolor_pecho")
```

```python
def calcular_riesgo_paciente(paciente, niveles_riesgo):
  return sum(niveles_riesgo.get(s, 0) for s in paciente["sintomas"]) # En lugar de crear lista temporal, la suma se realiza directamente.

print(calcular_riesgo_paciente(p, niveles_riesgo))
```

```python
def procesar_pacientes(pacientes_origen, niveles_riesgo):
  resultado = []
  for text in pacientes_origen:
    paciente = parsear_paciente(text)
    paciente["riesgo_total"] = calcular_riesgo_paciente(paciente, niveles_riesgo)
    resultado.append(paciente)
  return resultado

procesados = procesar_pacientes(pacientes, niveles_riesgo)
print(procesados[0])
```

```python
def ordenar_por_riesgo(pacientes_procesados):
  return sorted(pacientes_procesados, key=lambda p: p["riesgo_total"], reverse=True)
```

```python
import pandas as pd

def mostrar_resultados(pacientes_ordenados):
  df = pd.DataFrame(pacientes_ordenados)
  print(df[["nombre", "edad", "sintomas", "riesgo_total"]].to_string(index=False))
  return df
```

```python
def main_pacientes():
  procesados = procesar_pacientes(pacientes, niveles_riesgo)
  ordenados = ordenar_por_riesgo(procesados)
  mostrar_resultados(ordenados)

main_pacientes()
```

```python
clientes = [
    {"nombre": "Ana", "edad": 42, "codigo": 0},
    {"nombre": "Carlos", "edad": 30, "codigo": 2},
    {"nombre": "Isabel", "edad": 55, "codigo": 3},
    {"nombre": "Jorge", "edad": 40, "codigo": 1},
    {"nombre": "Marta", "edad": 28, "codigo": 0}
]

niveles_seguro = {
    0: {"categoria": "Riesgo Bajo", "precio_base": 120, "fraude": 0.01},
    1: {"categoria": "Riesgo medio", "precio_base": 200, "fraude": 0.05},
    2: {"categoria": "Riesgo alto", "precio_base": 350, "fraude": 0.15},
    3: {"categoria": "Riesgo crítico", "precio_base": 500, "fraude": 0.30}
}
```

```python
def clasificar_cliente(cliente, niveles_seguro):
  info = niveles_seguro[cliente["codigo"]]
  cliente_nuevo = cliente.copy()
  cliente_nuevo["categoria"] = info["categoria"]
  cliente_nuevo["precio_base"] = info["precio_base"]
  cliente_nuevo["fraude"] = info["fraude"]
  return cliente_nuevo

print(clasificar_cliente(clientes[0], niveles_seguro))
```

```python
def calcular_precio_final(cliente):
  cliente["precio_final"] = round(cliente["precio_base"] * (1 + cliente["edad"] / 100), 2)
  return cliente

c = clasificar_cliente(clientes[0], niveles_seguro)
print(calcular_precio_final(c))
```

```python
def procesar_clientes(clientes, niveles_seguro):
  resultado = []
  for cliente in clientes:
    c = clasificar_cliente(cliente, niveles_seguro)
    c = calcular_precio_final(c)
    resultado.append(c)
  return resultado

procesados = procesar_clientes(clientes, niveles_seguro)
print(procesados[0])
```

```python
import pandas as pd

def mostrar_resultados(clientes_procesados):
  df = pd.DataFrame(clientes_procesados)
  df_ordenado = df.sort_values(by=["fraude", "precio_final"], ascending=False)
  columnas = ["nombre", "edad", "categoria", "precio_base", "precio_final", "fraude"]
  print(df_ordenado[columnas].to_string(index=False))
  return df_ordenado

df = mostrar_resultados(procesados)
```

```python
def mostrar_analisis(clientes_procesados):
  max_fraude = max(clientes_procesados, key=lambda c: c["fraude"])
  max_precio = max(clientes_procesados, key=lambda c: c["precio_final"])
  prioritario = max(clientes_procesados, key=lambda c: c["fraude"] + c["precio_final"] / 1000)

  print("Análisis")
  print(f"Mayor probabilidad de fraude: {max_fraude["nombre"]} ({max_fraude["fraude"]})")
  print(f"Precio final más alto: {max_precio["nombre"]} ({max_precio["precio_final"]})")
  print(f"Revisar prioritariamente: {prioritario["nombre"]}")
```

```python
def main_seguros():
  procesados = procesar_clientes(clientes, niveles_seguro)
  df = mostrar_resultados(procesados)
  mostrar_analisis(procesados)

main_seguros()
```

```python
import string
import random

def obtener_caracteres(mayusculas=True, minusculas=True, numeros=True, signos=False):
  pool = ""   # string vacío
  if mayusculas: pool += string.ascii_uppercase
  if minusculas: pool += string.ascii_lowercase
  if numeros: pool += string.digits
  if signos: pool += string.punctuation
  return pool

print(obtener_caracteres())
print(obtener_caracteres(signos=True))
print(obtener_caracteres(minusculas=False, numeros=False))
```

```python
def garantizar_minimo(mayusculas=True, minusculas=True, numeros=True, signos=False):
  obligatorios = []
  if mayusculas: obligatorios.append(random.choice(string.ascii_uppercase))
  if minusculas: obligatorios.append(random.choice(string.ascii_lowercase))
  if numeros: obligatorios.append(random.choice(string.digits))
  if signos: obligatorios.append(random.choice(string.punctuation))
  return obligatorios

print(garantizar_minimo())
print(garantizar_minimo(signos=True))
```

```python
def crear_contrasena(longitud=12, mayusculas=True, minusculas=True, numeros=True, signos=False):
  pool = obtener_caracteres(mayusculas, minusculas, numeros, signos)
  contrasena = garantizar_minimo(mayusculas, minusculas, numeros, signos)

  resto = longitud - len(contrasena) # Diferencia para rellenar caracteres faltantes con aleatorios del pool
  contrasena += random.choices(pool, k=resto) # Une las dos listas

  random.shuffle(contrasena) # Mezcla los caracteres
  return "".join(contrasena) #

print(crear_contrasena(12))
print(crear_contrasena(16, signos=True))
print(crear_contrasena(8, mayusculas=False, signos=True))
```

```python
def main_contrasena():
  print("Generador de contraseñas")
  print(f"Estándar (12): {crear_contrasena(12)}")
  print(f"Con signos (16): {crear_contrasena(16, signos=True)}")
  print(f"Solo números y minúsculas (8): {crear_contrasena(8, mayusculas=False)}")
  print(f"Seguridad máxima (20): {crear_contrasena(20, signos=True)}")

main_contrasena()
```
